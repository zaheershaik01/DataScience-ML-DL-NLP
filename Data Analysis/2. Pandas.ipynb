### Pandas - DataFrame And Series

Pandas is a powerful data manipulation library in Python, widely used for data analysis and data cleaning. It provides two primary data structures: Series and DataFrame. A Series is a one-dimensional array-like object, while a DataFrame is a two-dimensional, size-mutable, and potentialially heterogenous tabular data structure with labeled axes (rows and columns).

```python
## Series
## A Pandas Series is a one-dimensional array-like object that can hold any data type. It is similar to a column in a table.

import pandas as pd
data = [1,2,3,4,5]
series = pd.Series(data)
print("Series \n", series)
print(type(series))
```

```python
## Create a Series from dictionary
data = {'a':1, 'b':2, 'c':3}
series_dict = pd.Series(data)
print(series_dict)
```

```python
data = [10, 20, 30]
index = ['a','b','c']
pd.Series(data, index=index)
```

```python
## DataFrame

data = {
    'Name':['Krish', 'John', 'Zaheer'],
    'Age':[25, 30, 45],
    'City':['Banglore', 'Hyderabad', 'Riyadh']
}
dff = pd.DataFrame(data)
print(dff)
print(type(dff))
```

```python
## Create a Data Frame From a List of Dictionaries

data = [
    {'Name': 'Zaheer', 'Age': 21, 'City': 'HYD'},
    {'Name': 'Hashid', 'Age': 20, 'City': 'KTDM'},
    {'Name': 'Asif', 'Age': 22, 'City': 'KMM'},
    {'Name': 'Rooman', 'Age': 21, 'City': 'WL'}
]
df = pd.DataFrame(data)
print(df)
print(type(df))
```
