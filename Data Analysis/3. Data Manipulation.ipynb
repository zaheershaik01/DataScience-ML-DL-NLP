### Data Manipulation and Analysis with Pandas

Data Manipulation and analysis are key tasks in any data science or data analysis project. Pandas provides a wide range of functions for data manipulation and analysis, making it easier to clean, transform, and extract insights from data. In this lesson, we will cover various data manipulation and analysis techniques using Pandas.

```python
import pandas as pd
```

```python
df = pd.read_csv('data.csv')
## Fetch the first 5 rows
df.head(5)
```

```python
df.tail(5)
```

```python
df.describe()
```

```python
df.dtypes
```

```python
## Handling Missing Values
df.isnull().any()
```

```python
df.isnull().sum()
```

```python
df_filled = df.fillna(0)
```

```python
### Filling missing values with the mean of the column
df['Calories_fillNA'] = df['Calories'].fillna(df['Calories'].mean())
df
```

```python
df.dtypes
```

```python
## Renaming Columns
df = df.rename(columns={'Total_duration':'Total Duration'})
df.head()
```

```python
## Change datatype
df['Calories_new'] = df['Calories'].fillna(df['Calories'].mean()).astype(int)
df.head()
```

```python
df['New Calories'] = df['Calories'].apply(lambda x:x*2)
df.head()
```

```python
## Data Aggregating and Grouping
df.head()
```

```python
grouped_mean = df.groupby('Total Duration')['Pulse'].mean()
print(grouped_mean)
```

```python
grouped_sum = df.groupby(['Total Duration', 'Maxpulse'])['Pulse'].sum()
print(grouped_sum)
```

```python
df.groupby(['Total Duration', 'Maxpulse'])['Pulse'].mean()
```

```python
## Aggregate multiple functions
grouped_agg = df.groupby('Total Duration')['Pulse'].agg(['mean','sum','count'])
grouped_agg
```

```python
### Merging and Joining DataFrames
# Create sample DataFrames
df1 = pd.DataFrame({'Key': ['A', 'B', 'C'], 'Value1': [1, 2, 3]})
df2 = pd.DataFrame({'Key': ['A', 'B', 'D'], 'Value2': [4, 5, 6]})
```

```python
df1
```

```python
df2
```

```python
## Merge DataFrames on the 'Key Columns' 
pd.merge(df1, df2, on="Key", how="inner")
```

```python
pd.merge(df1, df2, on="Key", how="outer")
```

```python
pd.merge(df1, df2, on="Key", how="left")
```

```python
pd.merge(df1, df2, on="Key", how="right")
```
